## 🔧 Parche de compatibilidad — `get_updates` eliminado en tf-keras moderno

El error `AttributeError: 'SGD' object has no attribute 'get_updates'` ocurre porque `model.py` de Mask R-CNN llama a `optimizer.get_updates()`, un método de Keras 2 que fue eliminado en `tf-keras` moderno.

La solución es parchear directamente la función `compile_model` en `mrcnn/model.py` para usar la API actual del optimizador.

```python
# CORRECCIÓN — Parchear model.py para eliminar la dependencia de get_updates()
import re

model_py_path = '/content/prueba1/mrcnn/model.py'

with open(model_py_path, 'r') as f:
    source = f.read()

# El problema está en la llamada a set_model y get_updates dentro de compile_model.
# Reemplazamos el bloque del optimizador legacy por la API moderna de tf-keras.
old_snippet = '''        self.keras_model.compile(
            optimizer=keras.optimizers.SGD(
                lr=learning_rate, momentum=config.LEARNING_MOMENTUM,
                clipnorm=config.GRADIENT_CLIP_NORM),
            loss=[None] * len(self.keras_model.outputs))'''

new_snippet = '''        # PATCH: lr → learning_rate (argumento renombrado en tf-keras moderno)
        self.keras_model.compile(
            optimizer=keras.optimizers.SGD(
                learning_rate=learning_rate, momentum=config.LEARNING_MOMENTUM,
                clipnorm=config.GRADIENT_CLIP_NORM),
            loss=[None] * len(self.keras_model.outputs))'''

if old_snippet in source:
    source = source.replace(old_snippet, new_snippet)
    print("✅ Parche 'lr → learning_rate' aplicado.")
else:
    print("⚠️  Snippet exacto no encontrado — buscando con regex...")
    # Fallback regex por si el formato de indentación difiere
    source, n = re.subn(
        r'(SGD\s*\()\s*lr\s*=',
        r'\1learning_rate=',
        source
    )
    print(f"   Reemplazos realizados: {n}")

# Segundo parche: el argumento 'lr' también aparece en LearningRateSchedule callbacks
source, n2 = re.subn(r'(?<![a-z_])lr=learning_rate', 'learning_rate=learning_rate', source)

with open(model_py_path, 'w') as f:
    f.write(source)

print("✅ model.py actualizado correctamente.")
```

```python
class CustomConfig(Config):
    """Configuración para el entrenamiento en el conjunto de datos de células sanguíneas (RBC / WBC)."""
    NAME = "object"
    GPU_COUNT = 1
    IMAGES_PER_GPU = 1
    NUM_CLASSES = 1 + 3  # Fondo + RBC + WBC + Platelets
    IMAGE_MIN_DIM = 480
    IMAGE_MAX_DIM = 640
    STEPS_PER_EPOCH = 500
    VALIDATION_STEPS = 5
    BACKBONE = 'resnet50'
    RPN_ANCHOR_SCALES = (8, 16, 32, 64, 128)
    TRAIN_ROIS_PER_IMAGE = 32
    MAX_GT_INSTANCES = 50
    POST_NMS_ROIS_INFERENCE = 500
    POST_NMS_ROIS_TRAINING = 1000

config = CustomConfig()
config.display()
```

```python
import xml.etree.ElementTree as ET
import skimage.io

class CustomDataset(utils.Dataset):

    def load_custom(self, dataset_dir, subset):
        """Carga el dataset buscando imágenes en 'imag' y etiquetas en 'label'."""
        self.add_class("object", 1, "RBC")
        self.add_class("object", 2, "WBC")
        self.add_class("object", 3, "Platelets")

        assert subset in ["train", "val"]
        subset_dir = os.path.join(dataset_dir, subset)
        image_dir = os.path.join(subset_dir, "imag")
        label_dir = os.path.join(subset_dir, "label")

        if not os.path.exists(image_dir):
            print(f"Error: No se encontró la carpeta {image_dir}")
            return

        for filename in os.listdir(image_dir):
            if filename.lower().endswith(('.jpg', '.jpeg', '.png')):
                image_path = os.path.join(image_dir, filename)
                ann_path   = os.path.join(label_dir, os.path.splitext(filename)[0] + '.xml')

                if os.path.exists(ann_path):
                    image = skimage.io.imread(image_path)
                    height, width = image.shape[:2]
                    self.add_image(
                        "object",
                        image_id=filename,
                        path=image_path,
                        width=width, height=height,
                        ann_path=ann_path)

    def load_mask(self, image_id):
        """Genera máscaras binarias a partir del archivo XML."""
        info = self.image_info[image_id]
        tree = ET.parse(info['ann_path'])
        root = tree.getroot()

        masks = []
        class_ids = []
        name_dict = {"RBC": 1, "WBC": 2, "Platelets": 3}

        for obj in root.findall('object'):
            name = obj.find('name').text
            if name not in name_dict:
                continue
            xmlbox = obj.find('bndbox')
            y1 = int(xmlbox.find('ymin').text)
            x1 = int(xmlbox.find('xmin').text)
            y2 = int(xmlbox.find('ymax').text)
            x2 = int(xmlbox.find('xmax').text)
            mask = np.zeros([info["height"], info["width"]], dtype=np.uint8)
            mask[y1:y2, x1:x2] = 1
            masks.append(mask)
            class_ids.append(name_dict[name])

        if not masks:
            return super(self.__class__, self).load_mask(image_id)

        # uint8, NO bool — evita errores de resize en el pipeline de máscaras
        masks = np.stack(masks, axis=2).astype(np.uint8)
        class_ids = np.array(class_ids, dtype=np.int32)
        return masks, class_ids

    def image_reference(self, image_id):
        return self.image_info[image_id]["path"]
```

```python
# PUNTO CRÍTICO 3 — Generar script externo de entrenamiento
# Incluye el mismo parche de model.py para garantizar sesión limpia
train_script = '''
import os
import sys
import re
import xml.etree.ElementTree as ET
import numpy as np
import skimage.io

os.environ['TF_USE_LEGACY_KERAS'] = '1'
import tensorflow as tf
tf.compat.v1.disable_eager_execution()

ROOT_DIR = '/content/prueba1'
sys.path.append(ROOT_DIR)

# ── Parche lr → learning_rate en model.py ──────────────────────────────────
model_py_path = os.path.join(ROOT_DIR, 'mrcnn', 'model.py')
with open(model_py_path, 'r') as f:
    src = f.read()
src_patched, n = re.subn(r'(?<![a-z_])lr=learning_rate', 'learning_rate=learning_rate', src)
src_patched, n2 = re.subn(r'(SGD\\s*\\()\\s*lr\\s*=', r'\\1learning_rate=', src_patched)
if n + n2 > 0:
    with open(model_py_path, 'w') as f:
        f.write(src_patched)
# ────────────────────────────────────────────────────────────────────────────

from mrcnn.config import Config
from mrcnn import model as modellib, utils

MODEL_DIR       = os.path.join(ROOT_DIR, "logs")
COCO_MODEL_PATH = os.path.join(ROOT_DIR, "mask_rcnn_coco.h5")


class CustomConfig(Config):
    NAME = "object"
    GPU_COUNT = 1
    IMAGES_PER_GPU = 1
    NUM_CLASSES = 1 + 3
    IMAGE_MIN_DIM = 480
    IMAGE_MAX_DIM = 640
    STEPS_PER_EPOCH = 100
    VALIDATION_STEPS = 5
    BACKBONE = 'resnet50'
    USE_MINI_MASK = False


class CustomDataset(utils.Dataset):
    def load_custom(self, dataset_dir, subset):
        self.add_class("object", 1, "RBC")
        self.add_class("object", 2, "WBC")
        self.add_class("object", 3, "Platelets")
        subset_dir = os.path.join(dataset_dir, subset)
        image_dir  = os.path.join(subset_dir, "imag")
        label_dir  = os.path.join(subset_dir, "label")
        for filename in os.listdir(image_dir):
            if filename.lower().endswith(('.jpg', '.jpeg', '.png')):
                img_path = os.path.join(image_dir, filename)
                ann_path = os.path.join(label_dir, os.path.splitext(filename)[0] + '.xml')
                if os.path.exists(ann_path):
                    image = skimage.io.imread(img_path)
                    height, width = image.shape[:2]
                    self.add_image("object", image_id=filename, path=img_path,
                                   width=width, height=height, ann_path=ann_path)

    def load_mask(self, image_id):
        info = self.image_info[image_id]
        tree = ET.parse(info['ann_path'])
        root = tree.getroot()
        masks, class_ids = [], []
        name_dict = {"RBC": 1, "WBC": 2, "Platelets": 3}
        for obj in root.findall('object'):
            name = obj.find('name').text
            if name not in name_dict:
                continue
            b  = obj.find('bndbox')
            y1 = int(b.find('ymin').text); x1 = int(b.find('xmin').text)
            y2 = int(b.find('ymax').text); x2 = int(b.find('xmax').text)
            m  = np.zeros([info["height"], info["width"]], dtype=np.uint8)
            m[y1:y2, x1:x2] = 1
            masks.append(m)
            class_ids.append(name_dict[name])
        return np.stack(masks, axis=2).astype(np.uint8), np.array(class_ids, dtype=np.int32)

    def image_reference(self, image_id):
        return self.image_info[image_id]["path"]


if __name__ == "__main__":
    config = CustomConfig()

    dataset_train = CustomDataset()
    dataset_train.load_custom("/content/prueba1/images", "train")
    dataset_train.prepare()

    dataset_val = CustomDataset()
    dataset_val.load_custom("/content/prueba1/images", "val")
    dataset_val.prepare()

    model = modellib.MaskRCNN(mode="training", config=config, model_dir=MODEL_DIR)
    model.load_weights(COCO_MODEL_PATH, by_name=True,
                       exclude=["mrcnn_class_logits", "mrcnn_bbox_fc",
                                "mrcnn_bbox", "mrcnn_mask"])

    print("Iniciando entrenamiento...")
    model.train(dataset_train, dataset_val,
                learning_rate=config.LEARNING_RATE,
                epochs=5,
                layers='heads')
    print("Entrenamiento completado.")
'''

with open('/content/prueba1/train_mrcnn.py', 'w') as f:
    f.write(train_script)

print("Script train_mrcnn.py generado correctamente.")
```
